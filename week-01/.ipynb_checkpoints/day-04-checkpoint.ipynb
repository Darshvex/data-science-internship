```python

import csv

# Step 1: Create marks.csv with 5 students
with open("marks.csv", "w", newline="") as file:
    writer = csv.writer(file)

    # Write column headings
    writer.writerow(["Name", "Marks"])

    # Write student details
    writer.writerow(["Amit", 80])
    writer.writerow(["Priya", 90])
    writer.writerow(["Rahul", 60])
    writer.writerow(["Sneha", 70])
    writer.writerow(["Neha", 100])

# Step 2: Read the CSV file
with open("marks.csv", "r", newline="") as file:
    reader = csv.reader(file)
    next(reader)  # Skip the heading row

    students = []

    for row in reader:
        name = row[0]
        marks = int(row[1])
        students.append([name, marks])

# Step 3: Calculate average marks
total = sum(student[1] for student in students)
average = total / len(students)

print("Average marks:", average)

# Step 4: Print students scoring above average
print("Students scoring above average:")

for student in students:
    if student[1] > average:
        print(student[0])
```

```python

# Step 1: Create helpers.py
with open("helpers.py", "w") as file:
    file.write("""
def average(marks):
    return sum(marks) / len(marks)


def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "F"
""")

# Step 2: Import and use the module
import helpers

marks = [80, 90, 70, 60, 100]

print("Average marks:", helpers.average(marks))
print("Grade:", helpers.get_grade(85))
```

```python

# Create a Student class
class Student:

    # Initialize student details
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Check whether the student has passed
    def is_pass(self):
        return self.marks >= 40

    # Display student details
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

        if self.is_pass():
            print("Result: Pass")
        else:
            print("Result: Fail")

        print()


# Create 3 student objects
s1 = Student("Amit", 80)
s2 = Student("Priya", 35)
s3 = Student("Rahul", 65)

# Display details of all students
s1.display()
s2.display()
s3.display()
```

```python

# Create a BankAccount class
class BankAccount:

    def __init__(self, balance=0):
        self.balance = balance

    # Deposit money
    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    # Withdraw money
    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient balance!")

        self.balance -= amount
        print("Withdrawn:", amount)

    # Display current balance
    def display_balance(self):
        print("Current balance:", self.balance)


# Create a bank account with 1000 rupees
account = BankAccount(1000)

try:
    account.deposit(500)
    account.withdraw(200)
    account.withdraw(2000)

except ValueError as error:
    print("Error:", error)

# Display the remaining balance
account.display_balance()
```
