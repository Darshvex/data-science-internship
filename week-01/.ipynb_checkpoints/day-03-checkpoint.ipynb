```python
student = {
    "name": "Rahul",
    "age": 20,
    "marks": 75
}

print("Student name:", student["name"])
print("Age:", student["age"])
print("Marks:", student["marks"])

student["city"] = "Pune"
student["marks"] = 90

print(student)
```

```python
sentence = "python is easy and python is useful"

words = sentence.split()
count = {}

for word in words:
    if word in count:
        count[word] = count[word] + 1
    else:
        count[word] = 1

print("Word counts:", count)
```

```python
python_students = {"Rahul", "Amit", "Priya", "Sneha"}
sql_students = {"Amit", "Priya", "Vijay", "Rohan"}

print("Students in both:", python_students & sql_students)

print("Only in Python:", python_students - sql_students)

print("Students in either:", python_students | sql_students)
```

```python
employees = [
    {"name": "Rahul", "dept": "IT", "salary": 60000},
    {"name": "Amit", "dept": "HR", "salary": 45000},
    {"name": "Priya", "dept": "Sales", "salary": 55000},
    {"name": "Sneha", "dept": "IT", "salary": 40000},
    {"name": "Vijay", "dept": "Finance", "salary": 70000}
]

total_salary = 0

for emp in employees:
    if emp["salary"] > 50000:
        print(emp["name"], emp["salary"])

    total_salary = total_salary + emp["salary"]

average = total_salary / len(employees)

print("Average salary:", average)
```

```python
employees = [
    {"name": "Rahul", "dept": "IT", "salary": 60000},
    {"name": "Amit", "dept": "HR", "salary": 45000},
    {"name": "Priya", "dept": "Sales", "salary": 55000},
    {"name": "Sneha", "dept": "IT", "salary": 40000},
    {"name": "Vijay", "dept": "Finance", "salary": 70000}
]

sorted_employees = sorted(employees, key=lambda emp: emp["salary"], reverse=True)

print("Top 3 employees:")

for emp in sorted_employees[:3]:
    print(emp["name"], emp["salary"])
```

```python
prices = [200, 400, 600, 800, 1000]

# Add 18% GST
gst_prices = list(map(lambda price: price * 1.18, prices))

# Keep prices above 500
result = list(filter(lambda price: price > 500, gst_prices))

print("Prices after GST:", gst_prices)
print("Prices above 500:", result)
```

```python

contacts = {}

while True:
    print("\n--- Contact Book ---")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Delete Contact")
    print("4. Show All Contacts")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter name: ")
        phone = input("Enter phone number: ")

        contacts[name] = phone
        print("Contact added successfully!")

    elif choice == "2":
        name = input("Enter name to search: ")

        if name in contacts:
            print("Phone number:", contacts[name])
        else:
            print("Contact not found!")

    elif choice == "3":
        name = input("Enter name to delete: ")

        if name in contacts:
            del contacts[name]
            print("Contact deleted successfully!")
        else:
            print("Contact not found!")

    elif choice == "4":
        print("All Contacts:", contacts)

    elif choice == "5":
        print("Thank you!")
        break

    else:
        print("Invalid choice! Try again.")
```
