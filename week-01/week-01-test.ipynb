```python
def calculate_marks(marks):
    minimum = min(marks)
    maximum = max(marks)
    average = sum(marks) / len(marks)

    if average >= 80:
        grade = "A"
    elif average >= 60:
        grade = "B"
    elif average >= 40:
        grade = "C"
    else:
        grade = "Fail"

    return {
        "minimum": minimum,
        "maximum": maximum,
        "average": average,
        "grade": grade
    }


# Test on two different lists
marks1 = [85, 90, 75, 80]
marks2 = [30, 45, 50, 35]

print(calculate_marks(marks1))
print(calculate_marks(marks2))
```

```python
import csv

# Create a CSV file
with open("products.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "price", "category"])

    writer.writerow(["Laptop", 50000, "Electronics"])
    writer.writerow(["Mouse", 500, "Electronics"])
    writer.writerow(["Keyboard", 1000, "Electronics"])
    writer.writerow(["Shirt", 800, "Clothing"])
    writer.writerow(["Jeans", 1500, "Clothing"])
    writer.writerow(["Shoes", 2000, "Clothing"])
    writer.writerow(["Book", 300, "Books"])
    writer.writerow(["Notebook", 100, "Books"])

# Read the CSV file and calculate total price
totals = {}

with open("products.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        category = row["category"]
        price = int(row["price"])

        if category in totals:
            totals[category] += price
        else:
            totals[category] = price

# Print total price for each category
for category, total in totals.items():
    print(category, ":", total)
```

```python
from collections import Counter

try:
    with open("data.txt", "w") as file:
        file.write("apple banana apple mango banana apple")

    with open("data.txt", "r") as file:
        text = file.read()

    words = text.lower().split()
    count = Counter(words)

    print("Three most common words:")
    print(count.most_common(3))

except FileNotFoundError:
    print("File does not exist.")
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def give_raise(self, percent):
        self.salary = self.salary + (self.salary * percent / 100)


# Create three employees
e1 = Employee("Rahul", 30000)
e2 = Employee("Priya", 40000)
e3 = Employee("Amit", 25000)

# Give a 10% raise to each employee
e1.give_raise(10)
e2.give_raise(10)
e3.give_raise(10)

# Store employees in a list
employees = [e1, e2, e3]

# Sort employees by salary
employees.sort(key=lambda emp: emp.salary)

# Print the result
for emp in employees:
    print(emp.name, emp.salary)
```
