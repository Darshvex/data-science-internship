```python
import numpy as np

# 1. Create a 1-D array of numbers from 1 to 20
arr1 = np.arange(1, 21)

# 2. Create a 5x5 array of zeros
arr2 = np.zeros((5, 5))

# 3. Create a 3x3 identity matrix
arr3 = np.eye(3)

# 4. Create 10 evenly spaced numbers between 0 and 1
arr4 = np.linspace(0, 1, 10)

# Print each array, its shape, and its dtype
arrays = [arr1, arr2, arr3, arr4]

for i, arr in enumerate(arrays, start=1):
    print("\nArray {i}:")
    print(arr)
    print("Shape:", arr.shape)
    print("Data type:", arr.dtype)
```

```python

import numpy as np

marks = np.array([78, 85, 92, 66, 88, 45, 73])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Position of Highest Mark:", np.argmax(marks))
```

```python

import numpy as np

marks = np.array([78, 85, 92, 66, 88, 45, 73])

avg = np.mean(marks)

print("Marks above average:", marks[marks > avg])

print("Students below 50:", np.sum(marks < 50))
```

```python

import numpy as np

a = np.arange(1, 17).reshape(4, 4)

print("Array:\n", a)
print("Second row:", a[1, :])
print("Last column:", a[:, 3])
print("Central 2x2 block:\n", a[1:3, 1:3])
print("Sum of each row:", np.sum(a, axis=1))
print("Sum of each column:", np.sum(a, axis=0))
```

```python

import numpy as np

celsius = np.array([0, 25, 37, 100])

fahrenheit = (celsius * 9/5) + 32

print("Fahrenheit:", fahrenheit)
```

```python

import numpy as np

np.random.seed(42)

a = np.random.randint(1, 101, 100)

print("Mean:", np.mean(a))
print("Values above 50:", np.sum(a > 50))

a = np.where(a < 20, 20, a)

print("Updated array:", a)
```

```python

import numpy as np
import time

# Using Python loop
start = time.time()
total = 0

for i in range(1000000):
    total += i * i

loop_time = time.time() - start

# Using NumPy
start = time.time()
a = np.arange(1000000)
total = np.sum(a * a)

numpy_time = time.time() - start

print("Loop time:", loop_time)
print("NumPy time:", numpy_time)

if loop_time > numpy_time:
    print("NumPy is faster")
else:
    print("Python loop is faster")
```

```python

import numpy as np

a = np.array([10, 20, 30, 40, 50])

# Normalize to range 0 to 1
normalized = (a - np.min(a)) / (np.max(a) - np.min(a))
print("Normalized:", normalized)

# Standardize using z-score
standardized = (a - np.mean(a)) / np.std(a)
print("Standardized:", standardized)
```
