```python
import pandas as pd
marks = pd.Series(
    [70, 60, 95, 35, 86],
    index=["darshan", "sarthak", "harshad", "jay", "aditya"]
)

print('Student Marks')
print(marks)
print('Top Scorer')
print(marks.idxmax())
print('Average Marks')
print(marks.mean())
print('Students who scored above 70')
print(marks[marks > 70])
```

```python
import pandas as pd

data = {
    "name": ['darshan','harish','ramesh','ganesh','abhi','rohit'],
    "department": ['accountant', 'clerk', 'manager', 'supervisor', 'HR', 'production head'],
    "salary": [20000, 15000, 35000, 25000, 30000, 32000],
    "experience": [2, 3, 5, 4, 1, 3]
}

df = pd.DataFrame(data)


print('Shape of DataFrame')
print(df.shape)

print('Column Names')
print(df.columns)

print('Data Types')
print(df.dtypes)
```

```python
import pandas as pd

data = {
    "name": ['Atul','aryan','samarth'],
    "department": ['IT', 'hr', 'sales'],
    "salary": [30000, 25000, 35000],
}

df = pd.DataFrame(data)
print(df)


df.to_csv("employees.csv", index=False)

df2 = pd.read_csv("employees.csv")

print("Original Shape:")
print(df.shape)

print("New Shape:")
print(df2.shape)

if df.shape == df2.shape:
    print("Shape is unchanged")
else:
    print("Shape is changed")
```

```python
import pandas as pd

data = {
    "name": ['Jay', 'Aditya', 'Riya'],
    "department": ['IT', 'HR', 'Sales'],
    "salary": [40000, 28000, 32000]
}

df = pd.DataFrame(data)

df.to_csv("employees.csv", index=False)

df = pd.read_csv("employees.csv")

print("displaying the first  rows of dataset:")
print(df.head())

print("displaying the last  rows of dataset:")
print(df.tail(3))

print("displaying the shape of dataset:")
print(df.shape)

print("displaying the information of dataset:")
print(df.info())

print("displaying the describe of dataset:")
print(df.describe())
```

```python
import pandas as pd

df = pd.read_csv("employees.csv")

print("Missing Values:")
print(df.isnull().sum())

print("Unique Values:")
print(df.nunique())

print(df["name"].value_counts())
print(df["department"].value_counts())
```

```python
import pandas as pd

df = pd.read_csv("employees.csv")

random_rows = df.sample(n=3, random_state=42)

print(random_rows[['name', 'salary']])
```

```python
import pandas as pd

df = pd.read_csv("employees.csv")

df.to_excel("employees.xlsx", index=False)

df_excel = pd.read_excel("employees.xlsx")


print(df.equals(df_excel))
```
