```python
import pandas as pd

df = pd.read_csv("tips.csv")

a = df[df["total_bill"] > 40]
print("Bills above 40:")
print(a)
print("Count:", len(a))

b = df[df["day"].isin(["Sat", "Sun"]) & (df["time"] == "Dinner")]

print(b)
print("Count:", len(b))

c = df[(df["smoker"] == "Yes") & (df["tip"] > 5)]

print(c)
print("Count:", len(c))
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("tips.csv")

df["tip_percent"] = (df["tip"] / df["total_bill"] * 100).round(2)

df["bill_per_person"] = df["total_bill"] / df["size"]

df["size_group"] = np.where(df["size"] >= 4, "Large", "Small")

print(df.head())
```

```python
import pandas as pd

df = pd.read_csv("tips.csv")

def bill_category(bill):
    if bill < 15:
        return "Low"
    elif bill <= 30:
        return "Medium"
    else:
        return "High"

df["bill_category"] = df["total_bill"].apply(bill_category)

print(df["bill_category"].value_counts())
```

```python
import pandas as pd

df = pd.read_csv("tips.csv")

df["tip_percent"] = (df["tip"] / df["total_bill"] * 100).round(2)

df = df.rename(columns={
    "total_bill": "bill",
    "tip": "tip_amount"
})

df = df.drop(columns=["sex"])

high_tippers = df[df["tip_percent"] > 20].copy()

high_tippers.to_csv("high_tippers.csv", index=False)

print(high_tippers)
print("File saved successfully!")
```

```python
import pandas as pd

df = pd.read_csv("tips.csv")

df["details"] = df.apply(
    lambda row: f"{row['day']} | {row['time']} | {row['size']} people",
    axis=1
)

df["sex"] = df["sex"].map({
    "Male": "M",
    "Female": "F"
})

print(df.head())
```
